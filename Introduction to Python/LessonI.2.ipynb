### Help command in python
- `help()` an interactive built-in help system of python

```python
help()
```

- can also be used to look for documentation of specified Python code

```python
help(int)
```

- `?` and `??` may also be used

```python
?int
```

```python
int?
```

```python
int??
```

For more information, read the [Python 3.12.5 documentation](https://docs.python.org/3/).
