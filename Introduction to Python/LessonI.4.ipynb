### Magic commads
- commands that provide shell-like syntax
- `%` acts on one line
- `%%` for multiple lines

| Magic Method               | Description                                                  |
|----------------------------|--------------------------------------------------------------|
| %alias                     | define alias for system command                              |
| %alias_magic               | create alias for existing line                               |
| %bookmark \<name\>         | set bookmark to current directory                            |
| %bookmark \<name\> \<dir\> | set bookmark to \<dir\>                                      |
| %bookmark -l               | list all bookmarks                                           |
| %bookmark -d \<name\>      | remove bookmark                                              |
| %bookmark - r              | remove all bookmarks                                         |
| %cd                        | change working directory                                     |
| %edit                      | bring up an editor and execute code                          |
| %env                       | get, set, or list environment variables                      |
| %history                   | print input history                                          |
| %load                      | load code into current frontend                              |
| %lsmagic                   | list available magic functions                               |
| %matplotlib                | set up matplotlib to work interactively                      |
| %pinfo                     | detailed information about the object                        |
| %pip                       | run pip package manager                                      |
| %precision                 | set floating point precision                                 |
| %pwd                       | return the current working directory path                    |
| %pylab                     | load numpy and matplotlib                                    |
| %quickref                  | quick reference list                                         |
| %run                       | run the named file inside IPython as a program               |
| %save                      | save set of lines to a given filename                        |
| %store                     | store variables in IPython database                          |
| %timeit                    | time execution of Python expression                          |
| %who                       | print all interactive variables with some minimal formatting |
| %who_ls                    | return a sorted list of all interactive variables            |
| %whos                      | print all interactive variables with additional detials      |
| %writefile                 | writes the contents of a cell to a file                      |
| %%html                     | render cell as a block of HTML                               |
| %%javascript or %js        | run cell block of Javascript code                            |
| %%latex                    | render cell as a block of LaTeX                              |

 For complete listof Magic commands, see [Built-in magic commands](https://ipython.readthedocs.io/en/stable/interactive/magics.html)
