- powerful interactive Python shell

### Jupyter
- web-based notebook authoring and editing applications
- provides tools (and standards) for interactive computing with computational notebooks

For more details, see [Project Jupyter Documentation](https://docs.jupyter.org/en/latest/) and [Jupyter Widgets](https://github.com/jupyter-widgets)
